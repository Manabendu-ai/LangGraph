### **Prompts In LangGraph**

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
model = ChatGoogleGenerativeAI(
    model = "gemini-3.6-flash"
)


print(f"[INFO] Model : {model.model} Loaded Successfully!")
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are helpful assistant that helps to create and curate posts for a social media platform."),
        ("user", "I want to create a post about the topic {topic}")
    ]
)
```

```python
prompt.invoke(
    {
        "topic" : "Importance of AOP"
    }
)
```
